```python
# Task 2.1
import random

def task2_1(filename, quantity, maximum):
    # Open the file in write mode
    with open(filename, "w") as file:
        #Generate random numbers
        for i in range(quantity):
            number = random.randint(0, maximum)
            file.write(str(number) + "\n")

task2_1("randomnumbers_YourName, CentreNumber_IndexNumber.txt", 1000, 5000)
```

```python
# Task 2.2
def task2_2(list_of_integers):

    # Base case
    if len(list_of_integers) <= 1:
        return list_of_integers

    # Split the list into two halves
    middle = len(list_of_integers) // 2
    left = task2_2(list_of_integers[:middle])
    right = task2_2(list_of_integers[middle:])

    # Merge the two sorted halves
    sorted_list = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    # Add any remaining elements
    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])

    return sorted_list

    print(task2_2([56, 25, 4, 98, 0, 18, 4, 5, 7, 0]) == [0, 0, 4, 4, 5, 7, 18, 25, 56, 98])
```

```python
# Task 2.3

def task2_3(filename_in, filename_out):
    with open(filename_in, "r") as file:
        numbers = []
        for line in file:
            numbers.append(int(line.strip()))

    sorted_numbers = task2_2(numbers)

    with open(filename_out, "w") as file:
        for number in sorted_numbers:
            file.write(str(number) + "\n")
            
# Sort the file
task2_3(
    "randomnumbers_YourName_CentreNumber_IndexNumber.txt",
    "sortednumbers_YourName_CentreNumber_IndexNumber.txt"
)
```
